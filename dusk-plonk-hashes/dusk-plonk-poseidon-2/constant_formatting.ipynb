```python
def int_to_4_u64(x: int):
    mask = (1 << 64) - 1
    return [
        (x >> (64 * 0)) & mask,
        (x >> (64 * 1)) & mask,
        (x >> (64 * 2)) & mask,
        (x >> (64 * 3)) & mask,
    ]
```
