```python
from sage.all import * 
from Crypto.Hash import SHAKE256
```

```python
def int_to_4_u64(x: int):
    mask = (1 << 64) - 1
    return [
        (x >> (64 * 0)) & mask,
        (x >> (64 * 1)) & mask,
        (x >> (64 * 2)) & mask,
        (x >> (64 * 3)) & mask,
    ]
```

```python
def u_64_chunks_to_int(chunks):
    x = 0
    for el in reversed(chunks):
        x *= 2**64
        x += el
    return x
```

```python
def u8_to_u64(u8_list):
    assert len(u8_list) == 32
    
    result = []
    
    for i in range(0, 32, 8):
        value = 0
        for j in range(8):
            value |= u8_list[i + j] << (8 * j)  # little endian
        result.append(value)
        
    return result
```

```python
q = 0x73eda753299d7d483339d80809a1d80553bda402fffe5bfeffffffff00000001

n = 4

rounds = 64

byte_length = rounds * ceil(log(q, 2**8))
byte_length
```

```python
seed = ("GMiMC2_" + str(q) + "_n_" + str(n) + "_rounds_" + str(rounds)).encode('ascii')
shake = SHAKE256.new()
shake.update(seed)
```

```python
constants_raw = list(shake.read(byte_length))
constants = []

for r in range(0, rounds):
    c = u_64_chunks_to_int(u8_to_u64(constants_raw[r * ceil(log(q, 2**8)):(r + 1) * ceil(log(q, 2**8))]))
    c = c % q
    constants.append(int_to_4_u64(c))
```

```python
print("[")
for cnst in constants:
    txt = "["
    for c in cnst:
        txt += str(c) + ","
    txt += "],"
    print(txt)
print("]")
```
